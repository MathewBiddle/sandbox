```python
import requests
import json
import pandas as pd
```

```python
dois = [ '10.1371/journal.pone.0302497',
       '10.1016/j.dsr2.2024.105380',
       '10.1016/j.dsr2.2023.105335',
       '10.1016/j.marpolbul.2023.115840',
       '10.1016/j.dsr2.2023.105344',
       '10.1016/j.dsr2.2023.105321',
       '10.1111/fog.12655',
       '10.1578/am.49.6.2023.519',
       '10.1016/j.dsr2.2023.105322',
       '10.1016/j.dsr2.2023.105346',
       '10.1016/j.dsr2.2023.105341',
       '10.1126/sciadv.adl3169',
       '10.1007/s10236-024-01605-y',
       '10.1016/j.pocean.2024.103244',
       '10.1029/2024JC020993',
       '10.1016/j.pocean.2024.103263',
       '10.1016/j.coastaleng.2024.104617',
       '10.1016/j.eti.2024.103676',
       '10.1007/s10236-024-01608-9',
       '10.1016/j.dsr2.2024.105395',
       '10.3389/fmars.2024.1387180',
       '10.3389/fevo.2023.1288596',
       '10.1016/j.hal.2023.102522',
]
```

```python
base_url = "https://api.unpaywall.org/v2/doi/"

df = pd.DataFrame({'doi': dois})

for doi in df['doi']:

  url = f'{base_url}{doi}?email=someone@somewhere.com'

  try:
      response = requests.get(url)
      response.raise_for_status()  # Raise an exception for bad status codes
      data = response.json()
      open = data['is_oa']
      df.loc[df['doi']==doi,'Open'] = open
  except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```
